## Теоретический расчет

Область всех исходов: вся поверхность стены($S$).

Благоприятная область: поверхность мишени(равна половине площади стены = $\frac{S}{2}$).

Расчет вероятности:

Событие A - попадание дротика в мишень:

$P(A) = \frac{S/2}{S} = \frac{1}{2} = 0.5$

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle
import numpy as np
```

```python
def darts(size):
    x0, y0, R = 6, 6, 5
    side = np.sqrt(2 * np.pi * R**2)
    fig, ax = plt.subplots(figsize=(5,5))
    ax = plt.gca()
    ax.add_patch(Rectangle((0,0), side, side, facecolor='red'))
    ax.add_patch(Circle((x0, y0), R, facecolor='lightgreen'))

    X = np.random.uniform(0, side, size)
    Y = np.random.uniform(0, side, size)
    plt.scatter(X, Y, marker='x', c='black')

    inside = len([d for d in (X-x0)**2 + (Y - y0)**2 if d <= R**2])
    prob = inside/size

    plt.show()
    print(f'В мишень из {size} дротиков попало {inside}. Доля попавших составила {prob}')
```

```python
darts(100)
```
