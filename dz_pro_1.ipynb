# Задача

Во время сессии все студенты решили прийти к суровому преподавателю на экзамен с микронаушником. У каждого из них недалеко находится друг, который поможет списать. Но преподаватель знал, что такое может случиться, поэтому установил в квадратной аудитории (с длиной стороны N) глушилку, которая находится в центре стены, противоположной столу преподавателя. Радиус действия глушилки - половина длины стены аудитории. 

Все студенты с помощью жребия определяют рассадку. Найдите вероятность, случайно выбранный студент не сможет списать экзамен.

## Теоретический расчет

Область всех исходов: вся площадь аудитории($N^2$).

Благоприятная область: область действия глушилки(равна площади полукруга = $\frac{\pi \cdot (\frac{N}{2})^2}{2}$).

Расчет вероятности:

Событие A - Студент попадает в зону действия глушилки:

$P(A) = \frac{\pi \cdot \frac{N}{2}^2}{2 \cdot N^2} = \frac{\pi}{8} = 0.3927$

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle
import numpy as np
```

```python
def success(size):
    x0, y0, R = 3, 0, 3
    side = 6
    fig, ax = plt.subplots(figsize=(5,5))
    ax = plt.gca()
    ax.add_patch(Rectangle((0, 0), side, side, facecolor='red'))
    ax.add_patch(Circle((x0, y0), R, facecolor='lightgreen'))

    X = np.random.uniform(0, side, size)
    Y = np.random.uniform(0, side, size)
    plt.scatter(X, Y, marker='x', c='black')

    inside = np.sum((X - x0)**2 + (Y - y0)**2 <= R**2)
    prob = inside/size
    
    plt.xlim(0, side)
    plt.ylim(0, side)
    plt.show()
    print(f'В зону действия глушилки из {size} студентов попало {inside}. Доля попавших составила {prob}')
```

```python
success(1000)
```
