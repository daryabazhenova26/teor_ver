### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы? 

* Рассмотрим пункт $А$ – по каждой номинации остановлены различные призы

$P(A) = \overline{{A_{10}^5}} = {10^5} = 100 000$

```python
10**5
```

* Рассмотрим пункт $Б$ – по каждой номинации установлены одинаковые призы

$P(B) =  \overline{{С_{10}^5}} = {C_{10+5-1}^5} = {C_{14}^5} = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$

```python
round(sc.comb(14, 5))
```

### Задача 1.9
>Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза? 

* Каждое семизначное число отличается от друтого
порядком следования цифр (причем $n_1 = 3$, $n_2 = 2$, $n_3 = 2$, а их сумма равна 7), т.е. является перестановкой с повторениями из 7 элементов. Их число по формуле:

$P_{7}(3;2;2) = \frac{7!}{3! \cdot 2! \cdot 2!} = 210$

```python
round(
    sc.factorial(7)
    / (sc.factorial(3)
    * sc.factorial(2)
    * sc.factorial(2)),
    8,
)
```

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* а) Пусть событие А - получение слова «ТОР». Различные комбинации трех букв из имеющихся шести представляют размещения, так как могут отличаться как составом входящих
букв, так и порядком их следования (или и тем и другим), т.е. общее число случаев $n = A_{6}^3$: , из которых благоприятствует событию А $m = 1$ случай. По формуле:

$P(A) = \frac{m}{n} = \frac{1}{A_{6}^3} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
round(1/sc.perm(6, 3),
      8,
)
```

* Пусть событие В - получение слова "ТЕОРИЯ". Различные комбинации шести букв из имеющихся шести представляют собой перестановки, так как отличаются только порядком следования букв; т.е. общее число случаев $n = P_{6} = 6!$, из которых благоприятствует событию В $m = 1$ случай. Поэтому

$P(B) = \frac{m}{n} = \frac{1}{P_{6}} = \frac{1}{6!} = \frac{1}{720}$

```python
round(1/sc.perm(6, 6),
      8,
)
```

### Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С. 

* Пусть событие В - получение слова «АНАНАС». Так же, как и в примере $1.10$б, общее число случаев $n = P_6 = 6!$, но теперь число случаев $m$, благоприятствующих событию В, существенно больше, так как перестановка трех букв А, осуществляемая $Р_3 = 3!$ способами, и перестановка двух букв Н ($Р_2 = 2!$ способами) не меняет собранное из карточек слово «АНАНАС»; по правилу произведения (см. § 1.5) $m = P_3 \cdot P_2$.
Итак, 


$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_{6}} = \frac{3! \cdot 2!}{6!} = \frac{1}{60}$

```python
n = sc.perm(6, 6)
m = sc.perm(3, 3) *  sc.perm(2, 2)

p_b = m / n 
p_b
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар. 

* а) Пусть событие А - никакие два лица одного пола не сядут рядом. Общее число способов рассадки 10 лиц на 10 местах определяется числом перестановок $n = P_10 = 10!$. Если женщины займут четные места $5!$ способами, то мужчины будут занимать нечетные места также $5!$ способами, и наоборот, т.е. число случаев, благоприятствующих событию А, равно $m_1 = 2(5!)^2$. Итак, 


$P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3 628 000} = 0,00794$

```python
n = sc.perm(10, 10)
m1 = 2 * sc.perm(5, 5)** 2
p_a = m1 / n
p_a
```

* Пусть событие В - супруги, образующие пять супружеских пар, сядут рядом. Теперь число случаев $m_2$, благоприятствующих событию В, определяется числом $5!$ всевозможных перестановок 5 супружеских пар, причем в каждой паре возможна перестановка мужа и жены; по правилу произведения $m_2 = 5!2^5$. Итак, 


$P(B) = \frac{m_2}{n} = \frac{5(2!)^5}{10!} = \frac{120 \cdot 32}{3 628 000} = 0,00106$

```python
m2 =  sc.perm(5, 5) * 2**5 
p_b = m2 / n
p_b
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе. 

* а) Пусть событие А - пассажиры попали в два купе. Общее число способов выбора 7 любых мест из имеющихся в вагоне 36 определяется числом сочетаний $n = C_{36}^7$. Для нахождения числа $m_1$ случаев, благоприятствующих событию А, учтем, что 2 купе из 9 можно выбрать $C_{9}^2$ способами, а 7 мест из имеющихся в двух купе 8 мест - $C_{8}^7$; способами. По правилу произведения $m_1 = C_{36}^2 \cdot C_{8}^7$. Итак, 

$P(A) = \frac{m_1}{n} = \frac{C_{9}^2 \cdot C_{8}^7}{C_{36}^7} = \frac{36 \cdot 8}{8 347 680} = 0,0000345$

```python
n = sc.comb(36, 7)
m1 = sc.comb(9, 2) * sc.comb(8, 7)
p_a = m1 / n
p_a
```

* б) Пусть событие В - пассажиры попали в семь купе. 7 купе из 9 можно выбрать $C_{9}^7$ способами. Семь мест в семи купе можно получить, если в каждом купе выбрать по одному месту из четырех, что возможно $4^7$ способами. Общее число случаев, благоприятствующих
событию В, по правилу произведения равно $m_2 = C_{9}^7 \cdot 4^7$. Итак, 


$P(B) = \frac{m_2}{n} = \frac{C_{9}^7 \cdot 4^7}{C_{36}^7} = \frac{36 \cdot 16 384}{8 347 680} = 0,07066$

```python
n = sc.comb(36, 7)
m2 = sc.comb(9, 7)* 4**7
p_b = m2 / n
p_b
```

* в) Пусть событие D - пассажиры попали в три купе. 3 купе из 9 можно выбрать $C_{9}^3$ способами, а число способов выбора семи мест из 12 в трех купе определяется сложнее, чем в п. а) и б). Действительно, возможные варианты выбора 7 мест из 12 в трех купе следующие: 4 + 2 + 1,  3 + 3 + 1,  3 + 2 + 2, а за счет перестановок купе
таких вариантов будет соответственно 6, 3 и 3. Каждый из этих вариантов по правилу произведения может быть получен $C_{4}^4C_{4}^2C_{4}^1$, $C_{4}^3C_{4}^3C_{4}^1$, $C_{4}^3C_{4}^2C_{4}^2$ способами соответственно. В результате общее число случаев, благоприятствующих событию D, равно $m_3 = C_{9}^3(6C_{4}^4C_{4}^2C_{4}^1 + 3C_{4}^3C_{4}^3C_{4}^1 + 3C_{4}^3C_{4}^2C_{4}^2 )= 84(6 \cdot 24 + 3 \cdot 64 + 3 \cdot 144) = 84 \cdot 768 = 64 512)$. Итак,

$P(D) = \frac{m_3}{n} = \frac{64 512}{8 347 680} = 0,00773$

```python
m3 = sc.comb(9, 3) * (6 * sc.comb(4, 4)* sc.comb(4, 2)*sc.comb(4, 1) + 
        3 * sc.comb(4, 3)*sc.comb(4, 3)*sc.comb(4, 1)  + 
        3 * sc.comb(4, 3)*sc.comb(4, 2)*sc.comb(4, 2))
p_d = m3 / n
p_d
```
